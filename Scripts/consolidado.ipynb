```python
# Célula do Notebook: Declaração da Query CTAS de Consolidação Trienal (Athena)

query_athena_consolidada = """
CREATE TABLE tb_estruturada_consolidada
WITH (
    external_location = 's3://lab-980861984629/input/gold/consolidada/',
    format = 'PARQUET',
    bucketed_by = ARRAY['ano_pesquisa'],
    bucket_count = 1
) AS
SELECT 2023 AS ano_pesquisa, t.* FROM tb_estruturada_2023 t
UNION ALL
SELECT 2024 AS ano_pesquisa, t.* FROM tb_estruturada_2024 t
UNION ALL
SELECT 2025 AS ano_pesquisa, t.* FROM tb_estruturada_2025 t;
"""

print(query_athena_consolidada)
```
